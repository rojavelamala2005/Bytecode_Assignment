```python
# Print all keys
dict1 = {"name":"roja","age":22,"branch":"cse"}
for key in dict1.keys():
       print(key)
```

```python
# Print all values
dict1 = {"name":"roja","age":22,"branch":"cse"}
for value in dict1.values():
       print(value)
```

```python
# Print key-value pairs
dict1 = {"name":"roja","age":22,"branch":"cse"}
for key,value in dict1.items():
       print(key,value)
```

```python
# Calculate total sales
sales = [1000, 2000, 1500, 3000]

total = 0

for sale in sales:
    total = total + sale

print("Total Sales:", total)
```

```python
# Find highest sales value
sales = [1000, 2000, 1500, 3000]

highest = sales[0]

for sale in sales:
    if sale > highest:
        highest = sale

print("Highest Sales Value:", highest)
```

```python
# Count customers with purchases above 1000
purchases = [500, 1200, 800, 2500, 1500]

count = 0

for amount in purchases:
    if amount > 1000:
        count = count + 1

print("Customers with purchases above 1000:", count)
```

```python
# Print products with price greater than 500
products = {
    "Laptop": 50000,
    "Mouse": 300,
    "Keyboard": 800,
    "Pen": 20
}

for product in products:
    if products[product] > 500:
        print(product, products[product])
```
