 # 1.	Calculate Total Sales
sales = [1200, 1500, 1800, 2000, 1700]
Using a while loop, calculate the total sales.
________________________________________


```python
sales = [1200, 1500, 1800, 2000, 1700]
i = 0
total = 0
while i < len(sales):
      total = total+sales[i]
      i = i + 1
print(total)
```

6.	Find Average Sales
sales = [1000, 1200, 1500, 1800, 2000]
Calculate the average sales value.


```python
# 6. Find Average Sales
sales = [1000, 1200, 1500, 1800, 2000]

total = 0
i = 0
while i < len(sales):
    total += sales[i]
    i += 1

average = total / len(sales)
print("Average Sales:", average)
```

```python
# 16. Second Highest Sale
sales = [1200, 3000, 1800, 5000, 4500]

first = second = -1
i = 0

while i < len(sales):
    if sales[i] > first:
        second = first
        first = sales[i]
    elif sales[i] > second and sales[i] != first:
        second = sales[i]
    i += 1

print("Second Highest Sale:", second)
```

```python
# 20. Stop at Target Revenue
sales = [1000, 2000, 1500, 3000, 2500]

total = 0
count = 0
i = 0

while i < len(sales):
    total += sales[i]
    count += 1

    if total >= 5000:
        break

    i += 1

print("Revenue Achieved:", total)
print("Records Processed:", count)
```

```python
# 21. Build Frequency Dictionary
products = ["Laptop", "Mobile", "Laptop", "Tablet", "Mobile", "Laptop"]

freq = {}
i = 0

while i < len(products):
    if products[i] in freq:
        freq[products[i]] += 1
    else:
        freq[products[i]] = 1
    i += 1

print(freq)
```

```python
# 24. Monthly Growth Percentage
sales = [1000, 1200, 1500, 1800]

i = 1

while i < len(sales):
    growth = ((sales[i] - sales[i - 1]) / sales[i - 1]) * 100
    print(round(growth, 2), "%")
    i += 1
```

```python
# 25. Mini KPI Dashboard
sales = [1200, 1500, 1800, 2000, 1700]

total = 0
maximum = sales[0]
minimum = sales[0]
count = 0
i = 0

while i < len(sales):
    total += sales[i]

    if sales[i] > maximum:
        maximum = sales[i]

    if sales[i] < minimum:
        minimum = sales[i]

    count += 1
    i += 1

average = total / count

print("Total:", total)
print("Average:", average)
print("Maximum:", maximum)
print("Minimum:", minimum)
print("Records:", count)
```

```python
# 30. Sales Trend Analysis
sales = [1000, 1200, 1100, 1500, 1400]

increase = 0
decrease = 0
i = 1

while i < len(sales):
    if sales[i] > sales[i - 1]:
        increase += 1
    elif sales[i] < sales[i - 1]:
        decrease += 1

    i += 1

print("Increasing Months:", increase)
print("Decreasing Months:", decrease)
```
