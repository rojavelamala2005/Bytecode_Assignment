```python
january_customers = {101,102,103,104,105}
february_customers = {103,104,105,106,107}

repeat_customers = january_customers & february_customers
new_customers = february_customers - january_customers
only_january = january_customers - february_customers

print("Repeat Customers =", repeat_customers)
print("New Customers in February =", new_customers)
print("Only January Customers =", only_january)
```

```python
A = {1001,1002,1003,1004,1005}
B = {1004,1005,1006,1007}

both_campaigns = A & B
either_campaign = A | B
exactly_one = A ^ B

print("Both Campaigns =", both_campaigns)
print("Either Campaign =", either_campaign)
print("Exactly One Campaign =", exactly_one)
```

```python
day1 = {101,102,103,104,105}
day2 = {103,104,105,106,107}
day3 = {104,105,108}

all_days = day1 & day2 & day3
at_least_once = day1 | day2 | day3
only_day1 = day1 - day2 - day3

print("Visitors All 3 Days =", all_days)
print("Visitors At Least Once =", at_least_once)
print("Visitors Only Day1 =", only_day1)
```

```python
store1 = {"Laptop","Mouse","Keyboard","Monitor"}
store2 = {"Mouse","Monitor","Printer","Scanner"}

common_products = store1 & store2
exclusive_store1 = store1 - store2
exclusive_store2 = store2 - store1
total_unique = store1 | store2

print("Common Products =", common_products)
print("Exclusive Products Store1 =", exclusive_store1)
print("Exclusive Products Store2 =", exclusive_store2)
print("Total Unique Products =", total_unique)
print("Count =", len(total_unique))
```

```python
jan = {101,102,103,104,105}
feb = {103,104,105,106,107}
mar = {105,106,107,108,109}

total_unique = jan | feb | mar
all_months = jan & feb & mar

only_once = (jan - feb - mar) | (feb - jan - mar) | (mar - jan - feb)

jan_feb_not_mar = (jan & feb) - mar
feb_mar_not_jan = (feb & mar) - jan

print("Total Unique Customers =", total_unique)
print("Count =", len(total_unique))

print("Customers in All Months =", all_months)

print("Customers Targeted Only Once =", only_once)

print("January and February but Not March =", jan_feb_not_mar)

print("February and March but Not January =", feb_mar_not_jan)
```
